# Hawk Showcase

Demonstrates the new `Animal3D` API with the hawk skeleton:
- Loading from CSV
- Polygon sections and marker access
- Two transform modes (`transform_all` vs `transform_display_only`)
- **Mirroring**: passing 4 right-side markers auto-generates the left side
- Runtime marker exclusion
- Scaling and normalisation
- Plotting with style control

## 1. Loading a Hawk

```python
from morphing_birds import Animal3D, SkeletonDefinition, plot_plotly, plot_plotly_compare
import numpy as np

# Load a hawk from CSV — just two arguments
hawk = Animal3D("hawk", data="../data/mean_hawk_shape.csv")

print(f"Skeleton: {hawk.skeleton}")
print(f"Shape: {hawk.current_shape.shape}")
print(f"Analysis markers ({len(hawk.analysis_indices)}): {hawk.analysis_marker_names}")
print(f"Display-only markers ({len(hawk.display_only_indices)}): {hawk.skeleton.display_only_markers}")
```

## 2. Basic Plotting

```python
# Default plot — section styles come from the YAML config
fig = plot_plotly(hawk, colour='steelblue')
fig.show()
```

```python
# Clean figure with no axes — useful for presentations
fig = plot_plotly(hawk, colour='coral', axes_visible=False)
fig.show()
```

## 3. Two Transform Modes

### `transform_display_only` — for mocap data
When marker positions already contain rotations from motion capture, only the display-only markers (shoulders, hood, tailbase, tailpack) need rotating to match.

### `transform_all` — for visualisation
When the data has no rotation baked in, rotate the entire animal.

```python
# transform_display_only: only shoulders/hood/tailbase/tailpack rotate
# The wing and tail markers stay put (they already have the right rotation from mocap)
hawk_a = hawk.copy()
hawk_a.transform_display_only(bodypitch=30)
fig = plot_plotly(hawk_a, colour='steelblue')
fig.update_layout(title="transform_display_only(bodypitch=30)")
fig.show()
```

```python
# transform_all: the ENTIRE animal rotates together
hawk_b = hawk.copy()
hawk_b.transform_all(bodypitch=30)
fig = plot_plotly(hawk_b, colour='tomato')
fig.update_layout(title="transform_all(bodypitch=30)")
fig.show()
```

## 4. Mirroring: Pass 4 Right-Side Markers, Get a Full Bird

When you only have right-side data (4 markers for hawk: wingtip, primary, secondary, tailtip), passing them into `update_keypoints` automatically mirrors to create the left side. This works for both single frames and multi-frame animation data.

```python
# Create a hawk and get its right-side markers
hawk_mirror = hawk.copy()
right_only = hawk_mirror.right_markers  # shape (1, 4, 3)
print(f"Right-only shape: {right_only.shape}")

# Modify the right side — spread the wing wider
modified_right = right_only.copy()
modified_right[0, 0, 0] *= 1.3  # stretch wingtip outward
modified_right[0, 0, 1] += 0.05  # push wingtip forward

# Pass just 4 markers — left side is auto-generated by mirroring
hawk_mirror.update_keypoints(modified_right)

print(f"After update — full markers shape: {hawk_mirror.markers.shape}")
print(f"Left wingtip x: {hawk_mirror.left_markers[0, 0, 0]:.4f}")
print(f"Right wingtip x: {hawk_mirror.right_markers[0, 0, 0]:.4f}")
print(f"Symmetric? {np.isclose(hawk_mirror.left_markers[0, 0, 0], -hawk_mirror.right_markers[0, 0, 0])}")

fig = plot_plotly(hawk_mirror, colour='mediumseagreen')
fig.update_layout(title="Mirrored from 4 right-side markers")
fig.show()
```

```python
# Mirroring also works with multi-frame data (e.g. for animations)
n_frames = 10
right_frames = np.tile(hawk.right_markers, (n_frames, 1, 1))

# Simulate wing sweep — gradually spread wings
for i in range(n_frames):
    factor = 1.0 + 0.05 * i
    right_frames[i, 0, 0] *= factor  # wingtip x
    right_frames[i, 1, 0] *= factor  # primary x

print(f"Right-only frames: {right_frames.shape}")

# format_keypoint_frames auto-mirrors for animations
from morphing_birds.plotting.animation_frame_helpers import format_keypoint_frames
full_frames = format_keypoint_frames(hawk, right_frames)
print(f"After mirroring: {full_frames.shape}")
print(f"Frame 0 left wingtip x: {full_frames[0, 0, 0]:.4f}")
print(f"Frame 0 right wingtip x: {full_frames[0, 1, 0]:.4f}")
print(f"Frame 9 left wingtip x: {full_frames[9, 0, 0]:.4f}")
print(f"Frame 9 right wingtip x: {full_frames[9, 1, 0]:.4f}")
```

## 5. Runtime Marker Exclusion

Dynamically add or remove markers from the analysis set without reloading.

```python
hawk_excl = hawk.copy()

print(f"Before: {len(hawk_excl.analysis_indices)} analysis markers")
print(f"  Names: {hawk_excl.analysis_marker_names}")

# Exclude tail tips from analysis
hawk_excl.exclude_markers(["left_tailtip", "right_tailtip"])
print(f"\nAfter excluding tailtips: {len(hawk_excl.analysis_indices)} analysis markers")
print(f"  Names: {hawk_excl.analysis_marker_names}")

# The markers are still plotted — they're just excluded from analysis indexing
fig = plot_plotly(hawk_excl, colour='steelblue')
fig.update_layout(title="Tailtips excluded from analysis (still visible)")
fig.show()

# Re-include them
hawk_excl.include_markers(["left_tailtip", "right_tailtip"])
print(f"\nAfter re-including: {len(hawk_excl.analysis_indices)} analysis markers")
```

## 6. Scaling and Normalisation

```python
from morphing_birds import compute_wingspan

hawk_scaled = hawk.copy()
ws_before = compute_wingspan(hawk_scaled.current_shape, hawk_scaled.skeleton)
print(f"Wingspan before: {ws_before:.4f} m")

# Normalise by wingspan — makes wingspan = 1.0
hawk_scaled.set_scale(normalise_by="wingspan")
ws_after = compute_wingspan(hawk_scaled.current_shape, hawk_scaled.skeleton)
print(f"Wingspan after normalisation: {ws_after:.4f}")

# Unit conversion
hawk_mm = hawk.copy()
hawk_mm.set_scale(unit_from="m", unit_to="mm")
print(f"\nAfter m -> mm conversion:")
print(f"  Wingspan: {compute_wingspan(hawk_mm.current_shape, hawk_mm.skeleton):.1f} mm")
```

## 8. Animations

### 8a. Basic animation with 8 analysis markers

Create synthetic wing-flapping motion and animate it.

```python
from morphing_birds import animate_plotly

# Create 30 frames of synthetic wing-flapping motion
# Start from the mean shape's analysis markers (8 markers)
n_frames = 30
base_markers = hawk.markers[0]  # (8, 3) — the 8 analysis markers

frames_8 = np.zeros((n_frames, 8, 3))
for i in range(n_frames):
    phase = np.sin(2 * np.pi * i / n_frames)
    frame = base_markers.copy()
    
    # Spread/retract wings (x-axis)
    frame[0, 0] *= (1.0 + 0.15 * phase)  # left_wingtip
    frame[1, 0] *= (1.0 + 0.15 * phase)  # right_wingtip
    frame[2, 0] *= (1.0 + 0.10 * phase)  # left_primary
    frame[3, 0] *= (1.0 + 0.10 * phase)  # right_primary
    
    # Sweep tail
    frame[6, 1] += 0.02 * phase  # left_tailtip y
    frame[7, 1] += 0.02 * phase  # right_tailtip y
    
    frames_8[i] = frame

print(f"Animation frames shape: {frames_8.shape}")
print(f"  {n_frames} frames x 8 analysis markers x 3 coords")

fig = animate_plotly(hawk, frames_8, colour='steelblue')
fig.show()
```

### 8b. Animation from 4 right-side markers only (auto-mirrors)

Pass just the right side — the animation pipeline mirrors each frame automatically.

```python
# Extract just the 4 right-side markers from each frame
# The right markers are at indices 1, 3, 5, 7 in the 8-marker analysis array
# (interleaved: left_wt, RIGHT_wt, left_pr, RIGHT_pr, ...)
right_frames = frames_8[:, 1::2, :]  # every other marker starting from index 1
print(f"Right-only frames: {right_frames.shape}")
print(f"  {n_frames} frames x 4 right markers x 3 coords")

# animate_plotly detects 4 markers and mirrors to 8 automatically
fig = animate_plotly(hawk, right_frames, colour='mediumseagreen')
fig.update_layout(title="Animated from 4 right-side markers (auto-mirrored)")
fig.show()
```

### 8c. Animation with body pitch per frame

Each frame can have its own body pitch (and yaw, roll, translation). The display-only markers rotate to match.

```python
# Create per-frame body pitch — oscillates between -20 and +20 degrees
bodypitch_frames = 20 * np.sin(2 * np.pi * np.arange(n_frames) / n_frames)

fig = animate_plotly(
    hawk, frames_8,
    colour='steelblue',
    bodypitch_frames=bodypitch_frames,
)
fig.update_layout(title="Animation with oscillating body pitch")
fig.show()
```

## 9. Full Pipeline: Exclude Markers → PCA-Style Analysis → Animate Results

This shows the complete workflow when you want to exclude some markers from analysis (e.g. the tail), run PCA on the remaining markers, then animate the reconstructed results back on the full animal.

**Key concept**: When you exclude markers, the `analysis_indices` and `analysis_marker_names` change immediately. Data you extract via `.markers` will have the reduced shape. But when you feed results back via `update_keypoints`, you need to provide data matching the *current* analysis marker count.

### Step 1: Set up the animal and exclude markers

```python
# Start with a fresh hawk
hawk_pca = Animal3D("hawk", data="../data/mean_hawk_shape.csv")

print("=== Before exclusion ===")
print(f"All markers:      {hawk_pca.skeleton.n_markers} total in current_shape")
print(f"Analysis markers:  {len(hawk_pca.analysis_indices)}")
print(f"Analysis names:    {hawk_pca.analysis_marker_names}")
print(f"Analysis indices:  {hawk_pca.analysis_indices}")
print(f"Display-only:      {hawk_pca.skeleton.display_only_markers}")
print(f".markers shape:    {hawk_pca.markers.shape}")

# Exclude the tailtips from analysis
hawk_pca.exclude_markers(["left_tailtip", "right_tailtip"])

print("\n=== After excluding tailtips ===")
print(f"Analysis markers:  {len(hawk_pca.analysis_indices)}")
print(f"Analysis names:    {hawk_pca.analysis_marker_names}")
print(f"Analysis indices:  {hawk_pca.analysis_indices}")
print(f"Display-only:      {[hawk_pca.marker_name_at(i) for i in hawk_pca.display_only_indices]}")
print(f".markers shape:    {hawk_pca.markers.shape}")
print(f"\nNote: .markers now returns (1, 6, 3) — only the 6 wing markers")
print(f"The tailtips are still in current_shape but excluded from analysis")
```

### Step 2: Generate synthetic motion data and extract the analysis subset

In a real workflow, this would be loading motion capture data. Here we simulate 50 frames of wing motion. We extract only the analysis markers (6 wing markers, no tailtips) — this is the data matrix you'd feed into PCA.

```python
# Simulate 50 frames of wing motion (all 8 original analysis markers)
n_frames = 50
base_all8 = hawk.markers[0]  # (8, 3) from the ORIGINAL hawk (before exclusion)

motion_all8 = np.zeros((n_frames, 8, 3))
for i in range(n_frames):
    phase = np.sin(2 * np.pi * i / n_frames)
    frame = base_all8.copy()
    # Wing sweep
    frame[0, 0] *= (1.0 + 0.2 * phase)   # left_wingtip
    frame[1, 0] *= (1.0 + 0.2 * phase)   # right_wingtip
    frame[2, 0] *= (1.0 + 0.12 * phase)  # left_primary
    frame[3, 0] *= (1.0 + 0.12 * phase)  # right_primary
    frame[4, 0] *= (1.0 + 0.05 * phase)  # left_secondary
    frame[5, 0] *= (1.0 + 0.05 * phase)  # right_secondary
    # Tail spread
    frame[6, 0] *= (1.0 + 0.1 * phase)   # left_tailtip
    frame[7, 0] *= (1.0 + 0.1 * phase)   # right_tailtip
    motion_all8[i] = frame

# Now extract ONLY the analysis markers (tailtips excluded)
# The analysis indices into the 8-marker array are 0,1,2,3,4,5 (the 6 wing markers)
# Tailtips were at indices 6,7 in the original marker order
analysis_idx_in_8 = [0, 1, 2, 3, 4, 5]  # wingtips, primaries, secondaries
motion_analysis = motion_all8[:, analysis_idx_in_8, :]

print(f"Full motion data:     {motion_all8.shape}  (50 frames x 8 markers x 3)")
print(f"Analysis-only data:   {motion_analysis.shape}  (50 frames x 6 markers x 3)")
print(f"  These 6 markers:    {hawk_pca.analysis_marker_names}")

# This is your PCA input matrix
# Flatten each frame: (50, 6, 3) -> (50, 18)
pca_input = motion_analysis.reshape(n_frames, -1)
print(f"\nPCA input matrix:     {pca_input.shape}  (50 observations x 18 features)")

# The labels for PCA component heatmaps
labels = hawk_pca.analysis_marker_labels()
print(f"Heatmap row labels:   {labels}")
```

### Step 3: Run PCA (simplified demo)

We do a simple PCA here to show the shape of the data at each step. In practice you'd use `sklearn.decomposition.PCA` or BirdPCA.

```python
# Simple PCA on the 6-marker analysis data
mean_shape_flat = pca_input.mean(axis=0)  # (18,)
centred = pca_input - mean_shape_flat     # (50, 18)

# SVD
U, S, Vt = np.linalg.svd(centred, full_matrices=False)
scores = U * S  # (50, 18) — PC scores per frame

print(f"Mean shape:      {mean_shape_flat.shape}  (18 features = 6 markers x 3 coords)")
print(f"PC components:   {Vt.shape}  (18 components x 18 features)")
print(f"PC scores:       {scores.shape}  (50 frames x 18 scores)")
print(f"Variance explained by PC1: {(S[0]**2 / (S**2).sum()) * 100:.1f}%")
print(f"Variance explained by PC2: {(S[1]**2 / (S**2).sum()) * 100:.1f}%")
```

### Step 4: Reconstruct from PCA and animate

Reconstruct the motion from PC1 only (the dominant wing-sweep mode), reshape back to `(n_frames, 6, 3)`, and feed into the animation.

**The critical detail**: `update_keypoints` expects data matching the *current* analysis marker count. Since we excluded 2 tailtips, it now expects 6 markers per frame. The display-only markers (shoulders, hood, tailbase, tailpack, **and now tailtips**) stay at their default positions.

```python
# Reconstruct using only PC1
n_pcs = 1
reconstructed_flat = (scores[:, :n_pcs] @ Vt[:n_pcs, :]) + mean_shape_flat  # (50, 18)

# Reshape back to (n_frames, 6_markers, 3_coords)
reconstructed_markers = reconstructed_flat.reshape(n_frames, 6, 3)
print(f"Reconstructed shape: {reconstructed_markers.shape}")
print(f"  {n_frames} frames x {len(hawk_pca.analysis_marker_names)} analysis markers x 3 coords")
print(f"  Markers: {hawk_pca.analysis_marker_names}")

# Animate the PCA reconstruction
# hawk_pca has tailtips excluded, so animate_plotly expects 6 markers per frame
# The tailtips + shoulders + hood + tailbase stay fixed at their default positions
fig = animate_plotly(hawk_pca, reconstructed_markers, colour='steelblue')
fig.update_layout(title="PC1 reconstruction (6 wing markers, tailtips fixed at default)")
fig.show()
```

### Step 5: Compare with the original full motion

To compare, use a **separate** `Animal3D` instance that still has all 8 analysis markers. The PCA hawk (`hawk_pca`) has tailtips excluded, so it expects 6 markers — don't try to feed 8 markers back into it.

```python
# Use the ORIGINAL hawk (all 8 analysis markers) to animate the full motion
hawk_full = Animal3D("hawk", data="../data/mean_hawk_shape.csv")

fig = animate_plotly(hawk_full, motion_all8, colour='tomato')
fig.update_layout(title="Original motion — all 8 markers including tail")
fig.show()
```

### Summary of the pipeline shapes

| Step | Object | Data shape | What it represents |
|------|--------|-----------|-------------------|
| Raw motion | `hawk_full` (8 analysis) | `(50, 8, 3)` | 50 frames, all 8 markers |
| Exclude tailtips | `hawk_pca` (6 analysis) | — | `.markers` now returns `(1, 6, 3)` |
| Extract analysis only | `motion_analysis` | `(50, 6, 3)` | 50 frames, 6 wing markers |
| Flatten for PCA | `pca_input` | `(50, 18)` | 50 observations x 18 features |
| PCA scores | `scores` | `(50, 18)` | 50 frames x 18 PCs |
| Reconstruct (PC1) | `reconstructed_markers` | `(50, 6, 3)` | Back to 6 wing markers |
| Animate reconstruction | `hawk_pca` + `(50, 6, 3)` | — | Wings move, tailtips frozen at default |
| Compare with original | `hawk_full` + `(50, 8, 3)` | — | Separate instance, all 8 markers moving |

**Key rule**: The number of markers you pass to `update_keypoints` or `animate_plotly` must match the animal's current analysis marker count. Use separate `Animal3D` instances if you need different exclusion sets.

## 10. Comparison Animation (`animate_plotly_compare`)

Overlay two sets of keypoint frames in the same animation to compare them side-by-side. Each set gets its own colour. Useful for comparing original vs PCA-reconstructed motion, or two different flight sequences.

### 10a. Compare original motion vs a modified version

Two different wing-sweep amplitudes animated together — one in blue, one in red.

```python
from morphing_birds import animate_plotly_compare

# Create two versions of wing motion with different amplitudes
n_frames = 40
base = hawk.markers[0]  # (8, 3)

# Version A: gentle sweep
frames_gentle = np.zeros((n_frames, 8, 3))
for i in range(n_frames):
    phase = np.sin(2 * np.pi * i / n_frames)
    frame = base.copy()
    frame[0, 0] *= (1.0 + 0.08 * phase)  # left_wingtip
    frame[1, 0] *= (1.0 + 0.08 * phase)  # right_wingtip
    frame[2, 0] *= (1.0 + 0.05 * phase)  # left_primary
    frame[3, 0] *= (1.0 + 0.05 * phase)  # right_primary
    frames_gentle[i] = frame

# Version B: dramatic sweep
frames_dramatic = np.zeros((n_frames, 8, 3))
for i in range(n_frames):
    phase = np.sin(2 * np.pi * i / n_frames)
    frame = base.copy()
    frame[0, 0] *= (1.0 + 0.25 * phase)  # left_wingtip
    frame[1, 0] *= (1.0 + 0.25 * phase)  # right_wingtip
    frame[2, 0] *= (1.0 + 0.18 * phase)  # left_primary
    frame[3, 0] *= (1.0 + 0.18 * phase)  # right_primary
    frame[6, 0] *= (1.0 + 0.15 * phase)  # left_tailtip
    frame[7, 0] *= (1.0 + 0.15 * phase)  # right_tailtip
    frames_dramatic[i] = frame

# Animate both together
fig = animate_plotly_compare(
    hawk,
    keypoints_frames_list=[frames_gentle, frames_dramatic],
    colours=['steelblue', 'tomato'],
    alpha=0.3,
)
fig.update_layout(title="Comparison: gentle sweep (blue) vs dramatic sweep (red)")
fig.show()
```

### 10b. Compare original vs PCA reconstruction

This is the key use case: overlay the original full motion against a PCA reconstruction to see what information each PC captures.

Since the PCA reconstruction has 6 markers (tailtips excluded) and the original has 8, we need two separate `Animal3D` instances.

```python
# We already have from section 9:
#   motion_all8:           (50, 8, 3) — original full motion
#   reconstructed_markers: (50, 6, 3) — PCA reconstruction (wings only)
#
# To compare them we can't use animate_plotly_compare directly, because
# that function expects all keypoint sets to use the SAME animal instance.
# The two sets have different marker counts (8 vs 6).
#
# Instead, we pad the reconstruction back to 8 markers by inserting the
# mean tailtip positions where the excluded markers were.
# This is a general pattern: "fill excluded markers with their defaults".

n_pca_frames = motion_all8.shape[0]
mean_tailtips = hawk.markers[0, 6:8, :]  # (2, 3) — default tailtip positions

reconstructed_full = np.zeros((n_pca_frames, 8, 3))
reconstructed_full[:, :6, :] = reconstructed_markers       # wings from PCA
reconstructed_full[:, 6, :]  = mean_tailtips[0]             # left tailtip fixed
reconstructed_full[:, 7, :]  = mean_tailtips[1]             # right tailtip fixed

print(f"Original frames:       {motion_all8.shape}")
print(f"Reconstructed (padded): {reconstructed_full.shape}")

# Now both are (50, 8, 3) and we can compare them with the full hawk
fig = animate_plotly_compare(
    hawk_full,
    keypoints_frames_list=[motion_all8, reconstructed_full],
    colours=['tomato', 'steelblue'],
    alpha=0.3,
)
fig.update_layout(title="Original (red) vs PC1 reconstruction (blue) — tailtips fixed in blue")
fig.show()
```

### 10c. Compare with per-frame transforms

You can also provide per-frame transforms (pitch, yaw, translation) for each keypoint set independently. Here both sets get the same oscillating body pitch.

```python
# Per-frame pitch for both sets
pitch = 15 * np.sin(2 * np.pi * np.arange(n_frames) / n_frames)

fig = animate_plotly_compare(
    hawk_full,
    keypoints_frames_list=[frames_gentle, frames_dramatic],
    colours=['steelblue', 'tomato'],
    alpha=0.3,
    # Each list element corresponds to one keypoint set
    bodypitch_frames_list=[pitch, pitch],
)
fig.update_layout(title="Comparison with oscillating body pitch applied to both")
fig.show()
```
