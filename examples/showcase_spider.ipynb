# Spider Showcase

Demonstrates the spider skeleton:
- Dict-based laterality (legs 1-4 = right, legs 5-8 = left, rather than prefix-based)
- Plotting individual legs
- Body section structure
- Transforms

```python
from morphing_birds import Animal3D, plot_partial_plotly, plot_plotly
```

## 1. Loading and Basic Plot

```python
spider = Animal3D("spider", data="../data/mean_spider_shape_carolina.csv")

print(f"Total markers: {spider.skeleton.n_markers}")
print(f"Analysis markers: {len(spider.analysis_indices)}")
print(f"Display-only: {spider.skeleton.display_only_markers}")
print(f"\nBody sections: {list(spider.polygons.keys())}")

fig = plot_plotly(spider, colour="darkorange")
fig.update_layout(title="Spider — Carolina wolf spider")
fig.show()
```

## 2. Dict-Based Laterality

Spiders use number suffixes (1-4 = right, 5-8 = left) instead of the `left_`/`right_` prefix convention used by birds.

```python
skel = spider.skeleton

print(f"Laterality config: {skel.laterality}")
print(f"\nRight-side markers ({len(skel.get_right_markers())}):")
for m in skel.get_right_markers()[:8]:
    print(f"  {m}")
print("  ...")

print(f"\nLeft-side markers ({len(skel.get_left_markers())}):")
for m in skel.get_left_markers()[:8]:
    print(f"  {m}")
print("  ...")

print(f"\nCentre markers: {skel.get_centre_markers()}")

print("\nMarker pairs (right leg <-> left leg):")
for right, left in skel.get_marker_pairs()[:8]:
    print(f"  {left:25s} <-> {right}")
```

## 3. Plotting Individual Legs

Use `plot_partial_plotly` with `leg_number` to isolate a single leg.

```python
# Plot leg 1 (right front)
fig = plot_partial_plotly(spider, leg_number=1, colour="crimson", alpha=0.8)
fig.update_layout(title="Spider — Leg 1 (right front)")
fig.show()
```

```python
# Plot leg 5 (left back — mirror of leg 4)
fig = plot_partial_plotly(spider, leg_number=5, colour="royalblue", alpha=0.8)
fig.update_layout(title="Spider — Leg 5 (left back)")
fig.show()
```

## 4. Transforms — Rotating the Whole Spider

```python
# Rotate the whole spider for a different viewing angle
spider_rotated = spider.copy()
spider_rotated.transform_all(bodypitch=45, bodyyaw=30)

fig = plot_plotly(spider_rotated, colour="darkorange")
fig.update_layout(title="Spider — Rotated (pitch=45, yaw=30)")
fig.show()
```

## 5. Excluding Legs from Analysis

You can exclude specific legs from analysis while keeping them visible for plotting. Useful for studying specific leg subsets.

```python
spider_subset = spider.copy()

print(f"Before: {len(spider_subset.analysis_indices)} analysis markers")

# Exclude legs 3 and 6 (a right-left pair)
leg3_markers = [f"{part}3" for part in ["claw", "tibiametatarsus", "patella", "coxa"]]
leg6_markers = [f"{part}6" for part in ["claw", "tibiametatarsus", "patella", "coxa"]]
spider_subset.exclude_markers(leg3_markers + leg6_markers)

print(
    f"After excluding legs 3 & 6: {len(spider_subset.analysis_indices)} analysis markers"
)
print(f"Excluded markers: {leg3_markers + leg6_markers}")

# The excluded legs are still plotted (they're in the polygons) but won't be
# included in analysis indices — useful for PCA on a subset of legs
fig = plot_plotly(spider_subset, colour="darkorange")
fig.update_layout(title="Spider — Legs 3 & 6 excluded from analysis (still visible)")
fig.show()

# Re-include
spider_subset.include_markers(leg3_markers + leg6_markers)
print(f"\nAfter re-including: {len(spider_subset.analysis_indices)} analysis markers")
```
