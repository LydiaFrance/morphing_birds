# Kestrel Showcase

Demonstrates the kestrel skeleton:
- Lab code column mapping (`r_p1_3` -> `right_firstprimary_tip`)
- Display names for plots (`right_firstprimary_tip` -> `R 1st Primary Tip`)
- Detailed marker set (32 analysis markers) vs simple variant (8 markers)
- Plotting individual sections (alula, handwing)

```python
from morphing_birds import Animal3D, plot_partial_plotly, plot_plotly
```

## 1. Full Mode — Lab Code Column Mapping

The kestrel CSV uses short lab codes (`r_p1_3`, `l_sh`, `h1`). The YAML config maps these to readable canonical names automatically.

```python
kestrel = Animal3D("kestrel", data="../data/mean_kestrel_shape.csv")

print(f"Total markers: {kestrel.skeleton.n_markers}")
print(f"Analysis markers: {len(kestrel.analysis_indices)}")
print(f"Display-only: {kestrel.skeleton.display_only_markers}")
print(f"\nBody sections: {list(kestrel.polygons.keys())}")

# Show some column mapping examples
print("\nColumn mapping (canonical -> CSV lab code):")
for canonical, lab_code in list(kestrel.skeleton.column_mapping.items())[:8]:
    print(f"  {canonical:30s} <- {lab_code}")

fig = plot_plotly(kestrel, colour="teal")
fig.update_layout(title="Kestrel — Full mode (32 analysis + 3 display-only)")
fig.show()
```

## 2. Display Names

Kestrel markers have optional display names for cleaner plot labels and PCA heatmaps.

```python
# Display names provide clean labels for plots
skel = kestrel.skeleton
print("Display name examples:")
print(f"  right_firstprimary_tip -> {skel.get_display_name('right_firstprimary_tip')}")
print(f"  left_tail_tip          -> {skel.get_display_name('left_tail_tip')}")
print(
    f"  left_elbow             -> {skel.get_display_name('left_elbow')}"
)  # no display name, falls back

# Labels for PCA heatmaps
labels = kestrel.analysis_marker_labels()
print(f"\nPCA heatmap labels ({len(labels)} total):")
for label in labels[:12]:
    print(f"  {label}")
```

## 3. Simple Variant

```python
# Simple variant — comparable to hawk marker set
kestrel_simple = Animal3D(
    "kestrel", data="../data/mean_kestrel_shape.csv", variant="simple"
)

print(f"Simple analysis markers: {len(kestrel_simple.analysis_indices)}")
print(f"Simple body sections: {list(kestrel_simple.polygons.keys())}")

fig = plot_plotly(kestrel_simple, colour="darkorange")
fig.update_layout(title="Kestrel — Simple variant")
fig.show()
```

## 4. Plotting Individual Sections

```python
# Plot just the right handwing section
fig = plot_partial_plotly(
    kestrel, section_name="right_handwing", colour="teal", alpha=0.7
)
fig.update_layout(title="Kestrel — Right handwing only")
fig.show()
```

```python
# Plot just the right alula
fig = plot_partial_plotly(
    kestrel, section_name="right_alula", colour="crimson", alpha=0.8
)
fig.update_layout(title="Kestrel — Right alula only")
fig.show()
```
